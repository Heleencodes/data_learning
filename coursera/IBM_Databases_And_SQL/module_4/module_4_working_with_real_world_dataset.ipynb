### Problem 3

##### Which schools have highest Safety Score?


```python
%%sql
SELECT Name_of_School, Safety_Score
FROM CHICAGO_PUBLIC_SCHOOLS_DATA
WHERE Safety_Score = 99;

```

```python
%%sql
SELECT Name_of_School, Safety_Score
FROM CHICAGO_PUBLIC_SCHOOLS_DATA
WHERE Safety_Score = (
    SELECT MAX(Safety_Score)
    FROM CHICAGO_PUBLIC_SCHOOLS_DATA
);
```

### Problem 6

##### Now remove the '%' sign from the above result set for Average Student Attendance column


```python
%%sql
SELECT 
    Name_of_School,
    AVERAGE_STUDENT_ATTENDANCE,
    CAST(REPLACE(AVERAGE_STUDENT_ATTENDANCE, '%', '') AS INTEGER) 
        AS Attendance_Number
FROM CHICAGO_PUBLIC_SCHOOLS_DATA
ORDER BY Attendance_Number ASC
LIMIT 5;
```

```python
%%sql
SELECT 
    Name_of_School,
    REPLACE(AVERAGE_STUDENT_ATTENDANCE, '%', '') AS Attendance_No_Percent
FROM CHICAGO_PUBLIC_SCHOOLS_DATA
ORDER BY Attendance_No_Percent ASC
LIMIT 5;
```

### Problem 7

##### Which Schools have Average Student Attendance lower than 70%?


```python
%%sql
SELECT 
    Name_of_School,
    AVERAGE_STUDENT_ATTENDANCE,
    CAST(REPLACE(AVERAGE_STUDENT_ATTENDANCE, '%', '') AS INTEGER) AS Attendance_Number
FROM CHICAGO_PUBLIC_SCHOOLS_DATA
WHERE CAST(REPLACE(AVERAGE_STUDENT_ATTENDANCE, '%', '') AS INTEGER) < 70;
```

### Problem 12

##### Get the hardship index for the community area which has the highest value for College Enrollment


```python
%%sql
SELECT DISTINCT COMMUNITY_AREA_NUMBER, "Hardship Index"
FROM CHICAGO_PUBLIC_SCHOOLS_DATA
ORDER BY COMMUNITY_AREA_NUMBER;
```

```python
%%sql
SELECT CD."Hardship Index"
FROM CENSUS_DATA CD
JOIN CHICAGO_PUBLIC_SCHOOLS_DATA CPS
    ON CD.community_area_number = CPS.community_area_number
WHERE CPS.college_enrollment = (
    SELECT MAX(college_enrollment)
    FROM CHICAGO_PUBLIC_SCHOOLS_DATA
);

```
